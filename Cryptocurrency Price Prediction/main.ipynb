## Cryptocurrency Price Prediction with Machine Learning

```python
import pandas as pd
import yfinance as yf
import datetime
from datetime import date, timedelta
```

collecting the latest data of Bitcoin prices for the past 730 days :

```python
today = date.today()

d1 = today.strftime("%Y-%m-%d")
end_date = d1
d2 = date.today() - timedelta(days=730)
d2 = d2.strftime("%Y-%m-%d")
start_date = d2
```

```python
data = yf.download('BTC-USD', 
                      start=start_date, 
                      end=end_date, 
                      progress=False)
data["Date"] = data.index
```

```python
data = data[["Date", "Open", "High", "Low", "Close", "Adj Close", "Volume"]]
data.reset_index(drop=True, inplace=True)
data.head()
```

```python
data.shape
```

visualizing the change in bitcoin prices till today by using a candlestick chart:

```python
import plotly.graph_objects as go
figure = go.Figure(data=[go.Candlestick(x=data["Date"],
                                        open=data["Open"], 
                                        high=data["High"],
                                        low=data["Low"], 
                                        close=data["Close"])])
figure.update_layout(title = "Bitcoin Price Analysis", 
                     xaxis_rangeslider_visible=False)
figure.show()
```

```python
# The Close column in the dataset contains the values we need to predict.

correlation = data.corr()
print(correlation["Close"].sort_values(ascending=False))
```

- Predicting the future prices of cryptocurrency is based on the problem of Time series analysis.

The AutoTS library in Python is one of the best libraries for time series analysis.

```python
from autots import AutoTS
model = AutoTS(forecast_length=30, frequency='infer', ensemble='simple')
model = model.fit(data, date_col='Date', value_col='Close', id_col=None)
```

```python
prediction = model.predict()
forecast = prediction.forecast
print(forecast)
```
